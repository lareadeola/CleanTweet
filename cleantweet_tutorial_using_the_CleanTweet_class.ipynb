Import the cleantweet library using the recommended alis - clt

```python
import cleantweet as clt
```

Create an instance of the class - CleanTweet Class object

```python
text = clt.CleanTweet('mini corpus for testing.txt')
```

Call the clean() method to clean the data

```python
clean_text_data = text.clean()
```

```python
clean_text_data
```

```python
type(clean_text_data)
```

Using the word_count attribute - attributes can only be used on objects

```python
text.word_count
```

Using the remove_articles() method which removes all articles from the text - both definite and indefinite articles.

```python
no_articles = text.remove_articles()
```

```python
no_articles
```

Using the remove_conjunctions() from the text, which will remove conjuctions like and, but etc

```python
no_conjunctions = text.remove_conjunctions()
```

```python
no_conjunctions
```
